```python
with open("bag_of_words.txt") as f:
    vocabulary = f.readlines()
```

```python
vocabulary = [v.strip().lower() for v in vocabulary]
vocabulary
```

```python
import json
with open("top_k_importance_results_Llama-3.3-70B-Instruct_yes.json") as f:
    yes_words = json.load(f)
```

```python
yes_words_unigrams = [w[0].strip("' ,.\"") for w in yes_words["unigrams"] if w[1] > 0]
```

```python
set(yes_words_unigrams).intersection(set(vocabulary))
```

```python
import json
with open("top_k_importance_results_Llama-3.3-70B-Instruct_no.json") as f:
    no_words = json.load(f)
```

```python
no_words_unigrams = [w[0].strip("' ,.\"") for w in no_words["unigrams"] if w[1] < 0]
```

```python
set(no_words_unigrams).intersection(set(vocabulary))
```

```python
def words_to_latex_table(words, cols=5):
    rows = []
    for i in range(0, len(words), cols):
        row = words[i:i + cols]
        # Pad the last row if needed
        row += [""] * (cols - len(row))
        rows.append(" & ".join(row) + r" \\")

    col_spec = "|".join(["l"] * cols)
    table = [
        r"\begin{tabular}{" + col_spec + "}",
    ]
    for row in rows:
        table.append(row)
    table.append(r"\end{tabular}")

    return "\n".join(table)
```

```python
print(words_to_latex_table(yes_words_unigrams))
```

```python
print(words_to_latex_table(no_words_unigrams))
```

```python
yes_words_bigrams = [w[0].strip("' ,.\"") for w in yes_words["bigrams"] if w[1] < 0]
no_words_bigrams = [w[0].strip("' ,.\"") for w in no_words["bigrams"] if w[1] < 0]
```

```python
print(words_to_latex_table(yes_words_bigrams[:40], 4))
```

```python
print(words_to_latex_table(no_words_bigrams[:40], 4))
```
