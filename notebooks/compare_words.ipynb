```python
with open("bag_of_words.txt") as f:
    vocabulary = f.readlines()
```

```python
vocabulary = [v.strip().lower() for v in vocabulary]
vocabulary
```

```python
import json
with open("top_k_importance_results_Llama-3.1-8B_yes.json") as f:
    yes_words = json.load(f)
```

```python
yes_words_unigrams = [w[0].strip("' ,.\"") for w in yes_words["unigrams"] if w[1] > 0]
```

```python
set(yes_words_unigrams).intersection(set(vocabulary))
```

```python
import json
with open("top_k_importance_results_Llama-3.1-8B_no.json") as f:
    no_words = json.load(f)
```

```python
no_words_unigrams = [w[0].strip("' ,.\"") for w in no_words["unigrams"] if w[1] < 0]
```

```python
set(no_words_unigrams).intersection(set(vocabulary))
```
