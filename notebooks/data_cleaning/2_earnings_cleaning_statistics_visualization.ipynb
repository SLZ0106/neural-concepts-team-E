```python
import polars as pl
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
import seaborn as sns
```

```python
BASE_DIR = Path.cwd().parent.parent
GITHUB_DATA_PATH = BASE_DIR / "data" 
OUTPUTS_PATH = BASE_DIR / "outputs"
```

# Load data

```python
with open(GITHUB_DATA_PATH / "outputs/cleaning_metrics.json",'r') as f:
    df = json.load(f)
```

# Figures

```python
## Global matplotly options
mpl.rcParams['axes.spines.right'] = False
mpl.rcParams['axes.spines.top'] = False
```

## 1. Figure total calls per year

```python
plt.figure(figsize=(8, 6))
sns.barplot(x='year', y='total_calls', data=df, color='skyblue')
plt.title('Total Earnings calls per year')
plt.savefig(OUTPUTS_PATH / "figures/total_earnings_calls.png", dpi = 500)
plt.show()
```

## 2. Figure parsing success per year

```python
plt.figure(figsize=(8, 6))
sns.barplot(x='year', y='success_rate', data=df, color='skyblue')
plt.title('Parsing success (%) of calls per year')
plt.savefig(OUTPUTS_PATH / "figures/parsing_success.png", dpi = 500)
plt.show()
```
